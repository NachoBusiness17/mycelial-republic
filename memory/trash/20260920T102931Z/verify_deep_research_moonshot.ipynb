# The Deep-Research Moonshot — live demo

`mag/deep_research_moonshot.py` self-steals the **internal** synthesis pipeline (deep research = mem tools, never external web):
PRIOR (`research_lens`) + WITNESS (`roshomon`) + SHADOW (`shadow_work`) + KNOT (`lens_chain`) → **THE MARRIAGE** (stochastic + deterministic + corpus = one chain link) → route the novel seam to the swarm via `swarm_surface.route_novel`.

Deterministic + $0 core. Let's run it against our own corpus.


```python
import sys, json
from pathlib import Path

ROOT = Path(r"c:\Users\foste\Documents\projects\local_sovereign_agent")
sys.path.insert(0, str(ROOT))

from mag import deep_research_moonshot as drm

print("=== STATUS (the contract) ===")
print(json.dumps(drm.status(), indent=2, default=str))
```

```python
# STEP 1 — synthesize a deep-research question against the INTERNAL corpus (the marriage, $0)
syn = drm.synthesize(
    "what is the one seam the standing wave has not yet shaded toward the frontier"
)

print("=== SYNTHESIZE: per-stage honest flags ===")
for s in syn["stages"]:
    print(f"  [{s['name']:<8}] ok={s['ok']}  {s.get('error','')[:60]}")

print("\n=== THE MARRIAGE (stochastic + deterministic + corpus -> one chain link) ===")
print(json.dumps(syn["marriage"], indent=2, default=str))
```

```python
# STEP 2 — the full moonshot: synthesize + route the NOVEL seam to the swarm (route_novel)
mo = drm.moonshot(
    "what is the one mechanism by which a swarm becomes smarter than its parts",
    route=True,   # the one primitive: dedupe by anchor + enqueue to the swarm research queue
)

print("=== MOONSHOT: the routed novel seam ===")
print(json.dumps(mo["routed"], indent=2, default=str))
print("\n=== MOONSHOT: the synthesized invariant handed to the swarm ===")
print(mo["synthesis"]["invariant"])
print("\nshadow edge:", mo["synthesis"]["shadow_edge"])
```
